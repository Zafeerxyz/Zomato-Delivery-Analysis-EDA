```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("cleaned_zomato.csv")
```

```python
def haversine(restaurant_latitude,restaurant_longitude,delivery_location_latitude,delivery_location_longitude):
   #radius of earth
    R = 6371
    
    #convert degree into radians
    restaurant_latitude = np.radians(restaurant_latitude)
    restaurant_longitude = np.radians(restaurant_longitude)

    delivery_location_latitude = np.radians(delivery_location_latitude)
    delivery_location_longitude = np.radians(delivery_location_longitude)
    

    latitude_diff = restaurant_latitude - delivery_location_latitude
    longitude_diff = restaurant_longitude - delivery_location_longitude

    #haversine formula
    a = (np.sin(latitude_diff/2)**2 
         +np.cos(restaurant_latitude)
         *np.cos(delivery_location_latitude)*np.sin(longitude_diff/2)**2)

    #central angle
    c = 2* np.arcsin(np.sqrt(a))

    #distance
    distance = R * c
    
    return round(distance,1)

    
```

```python
df["distance_km"] = haversine(df["restaurant_latitude"],
                              df["restaurant_longitude"],
                              df["delivery_location_latitude"],
                              df["delivery_location_longitude"])
```

**The distance_km is calculated using the Haversine formula, which measures the distance between the restaurant and the delivery location. It provides an approximate value because latitude and longitude measure a straight-line distance , whereas real-world roads are not straight lines; therefore, it is an approximation rather than an exact route distance.**

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["day_of_the_week"] = df["order_date"].dt.day_name()
```

**Finding day of the week from ordered_date**

```python
df["is_weekend"] = np.where((df["day_of_the_week"]=="Saturday")|(df["day_of_the_week"]=="Sunday"), 1, 0)
```

**Checking if its a weekend or not for analyzing the performance at weeekend**

```python
df["rating_category"] = np.floor(df["delivery_person_ratings"])
```

```python
def get_time_of_the_day(time):
    if pd.isna(time):
        return np.nan
    elif 5 <= time < 12 :
        return "Morning"
    elif 12<= time < 17:
        return "Afternoon"
    elif 17 <= time < 21:
        return "Evening"
    else:
        return "Night"

df["time_of_the_day"] = df["time_ordered"].apply(get_time_of_the_day)
```

**Can understand which timing has more orders**

```python
def get_distance_bucket(dist):
    if pd.isna(dist):
        return np.nan
    elif dist < 5:
        return "Short"
    elif 5 <= dist < 10:
        return "Medium"
    elif 10 <= dist < 15:
        return "Long"
    else:
        return "Very Long"


df["distance_bucket"] = df["distance_km"].apply(get_distance_bucket)
```

**For finding range of deliveries**

```python
df["distance_km"] = round(df["distance_km"],1)
df
```

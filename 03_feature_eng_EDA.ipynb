```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("cleaned_zomato.csv")
```

```python
df.head()
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["day_of_the_week"] = df["order_date"].dt.day_name()
```

**Finding day of the week from ordered_date**

```python
df["is_weekend"] = np.where((df["day_of_the_week"]=="Saturday")|(df["day_of_the_week"]=="Sunday"), 1, 0)
```

**Checking if its a weekend or not for analyzing the performance at weeekend**

```python
df["rating_category"] = round(df["delivery_person_ratings"])
```

**standardizing the rating for comaprison**

```python
df
```

```python
def get_time_of_the_day(time):
    if pd.isna(time):
        return np.nan
    elif 5 <= time < 12 :
        return "Morning"
    elif 12<= time < 17:
        return "Afternoon"
    elif 17 <= time < 21:
        return "Evening"
    else:
        return "Night"

df["time_of_the_day"] = df["time_ordered"].apply(get_time_of_the_day)
```

**Can understand which timing has more orders**

```python
def get_distance_bucket(dist):
    if pd.isna(dist):
        return np.nan
    elif dist < 5:
        return "Short"
    elif 5 <= dist < 10:
        return "Medium"
    elif 10 <= dist < 15:
        return "Long"
    else:
        return "Very Long"


df["distance_bucket"] = df["distance_km"].apply(get_distance_bucket)
```

**For finding range of deliveries**
